```python
# === EXERCISE 1: Emergency Dispatch Record Search ===
LAST_NAME = "Mendoza"
STUDENT_ID = "TUPM-26-1250"
SEED_DIGIT = int(STUDENT_ID[-1]) # 0
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # 16

# "Syd Hartha" -> S(19)+y(25)+d(4)+H(8)+a(1)+r(18)+t(20)+h(8)+a(1) = 104
ARTIST_SUM = 104 

# 1. Generate the original and rotated incident records
original_records = [100 + i * 5 for i in range(25)]
# Using SEED_DIGIT + 8 to ensure an observable rotation
rotation_point = SEED_DIGIT + 8 
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

# Generate target using ID_SUM and ARTIST_SUM
target_incident = original_records[(ID_SUM + ARTIST_SUM) % len(original_records)]

# 2. Search for the target incident in rotated array (Modified Binary Search)
def search_rotated(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        
        comparisons += 1
        if data[low] <= data[mid]: # Left half is sorted
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else: # Right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

# We will reuse linear search for comparison
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

# 3, 4, 5. Report position, record comparisons, and compare with Linear Search
rot_index, rot_comp = search_rotated(rotated_records, target_incident)
lin_index, lin_comp = linear_search(rotated_records, target_incident)

print("=== ASSESSMENT DATA ===")
print("Original Dataset:", original_records)
print("Rotated Dataset: ", rotated_records)
print("Target Incident: ", target_incident)
print(f"\nRotated Binary Search -> Index: {rot_index}, Comparisons: {rot_comp}")
print(f"Linear Search         -> Index: {lin_index}, Comparisons: {lin_comp}")
print("\n=== TIME AND SPACE COMPLEXITY ===")
print("Rotated Search Time Complexity: O(log n)")
print("Linear Search Time Complexity: O(n)")
print("Space Complexity (both): O(1)")
```

```python
# === EXERCISE 2: University Enrollment Records ===
NAME_LENGTH = 7
ARTIST_LENGTH = 10 # "Syd Hartha"

# 1. Generate the two sorted enrollment datasets
# Dataset A: Ascending values using SEED_DIGIT (0) and ID_SUM (16)
dataset_a = sorted([1000 + SEED_DIGIT * 10 + i * ID_SUM for i in range(12)])
# Dataset B: Ascending values using ARTIST_LENGTH (10) and NAME_LENGTH (7)
dataset_b = sorted([1050 + ARTIST_LENGTH * 5 + i * NAME_LENGTH for i in range(15)])

# 2. Consolidate them into one ordered collection (Merge process)
def merge_sorted_datasets(a, b):
    merged = []
    i, j, comparisons = 0, 0, 0
    
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
            
    # Append any remaining elements (no comparisons needed)
    merged.extend(a[i:])
    merged.extend(b[j:])
    
    return merged, comparisons

consolidated, merge_comparisons = merge_sorted_datasets(dataset_a, dataset_b)

# 4. Verify the final order and number of records
is_sorted = all(consolidated[k] <= consolidated[k+1] for k in range(len(consolidated)-1))
initial_count = len(dataset_a) + len(dataset_b)
final_count = len(consolidated)

print("=== ASSESSMENT DATA ===")
print(f"Department A Dataset ({len(dataset_a)} records):", dataset_a)
print(f"Department B Dataset ({len(dataset_b)} records):", dataset_b)
print(f"\nConsolidated Dataset ({final_count} records):", consolidated)

print("\n=== VERIFICATION & METRICS ===")
print(f"Records before: {initial_count} | Records after: {final_count}")
print(f"Ordering Verification Passed: {is_sorted}")
print(f"Comparison Count: {merge_comparisons}")

print("\n=== TIME AND SPACE COMPLEXITY ===")
print("Time Complexity: O(n + m) -> Linear time to traverse both datasets")
print("Space Complexity: O(n + m) -> Linear space to store the consolidated list")
```

```python
# === EXERCISE 3: E-Commerce Order Processing ===
import time
import random

# Re-define sorting functions for standalone execution
def bubble_sort(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comps += 1
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                moves += 1
                swapped = True
        if not swapped: break
    return a, comps, moves

def selection_sort(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[minimum]: minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            moves += 1
    return a, comps, moves

def insertion_sort(data):
    a, comps, moves = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comps += 1
            if a[j] <= key: break
            a[j+1] = a[j]
            moves += 1
            j -= 1
        a[j+1] = key
    return a, comps, moves

# 1. Generate input conditions
random.seed(16) # Seeded with ID_SUM for reproducibility
base_orders = [random.randint(1000, 9999) for _ in range(25)]

conditions = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random (Unsorted)": base_orders.copy()
}

# 2, 3, 4, 5. Apply algorithms, record metrics, verify, and compare
print("=== ALGORITHM PERFORMANCE BY INPUT CONDITION ===")
print(f"{'Condition':<18} | {'Algorithm':<10} | {'Comps':<5} | {'Moves':<5} | {'Time (s)':<8} | {'Sorted?'}")
print("-" * 75)

for condition_name, dataset in conditions.items():
    for algo_name, sort_func in [("Bubble", bubble_sort), ("Selection", selection_sort), ("Insertion", insertion_sort)]:
        start_time = time.perf_counter()
        sorted_res, comparisons, movements = sort_func(dataset)
        exec_time = time.perf_counter() - start_time
        
        # Verify
        is_sorted = all(sorted_res[k] <= sorted_res[k+1] for k in range(len(sorted_res)-1))
```
